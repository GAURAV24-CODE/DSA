```python
# A variable is a name used to refer to a value.

name = "Gaurav"
age = 22
course = "MCA"

# Display variable values
print(name)
print(age)
print(course)
```

```python
# Valid variable names
student_name = "Gaurav"
age2 = 22
_total = 500

# Display values
print(student_name)
print(age2)
print(_total)

# Rules:
# 1. Cannot start with a number.
# 2. Cannot contain spaces.
# 3. Cannot use Python keywords.
# 4. Variable names are case-sensitive.
```

```python
# Python determines the data type at runtime.

x = 100
print(x, type(x))

# The same variable can refer to a different type of object.
x = "Python"
print(x, type(x))

x = 10.5
print(x, type(x))
```

```python
# Common built-in data types

age = 22                         # int
price = 99.5                     # float
name = "Gaurav"                  # str
is_student = True                # bool
marks = [80, 90, 85]             # list
coordinates = (10, 20)           # tuple
numbers = {1, 2, 3}              # set
student = {"name": "Gaurav"}     # dict

# Display data types
print(type(age))
print(type(price))
print(type(name))
print(type(is_student))
print(type(marks))
print(type(coordinates))
print(type(numbers))
print(type(student))
```

```python
# Convert a string into an integer
age = int("22")
print(age, type(age))

# Convert an integer into a float
price = float(100)
print(price, type(price))

# Convert an integer into a string
marks = str(85)
print(marks, type(marks))
```

```python
# Store marks for five subjects
maths = 80
science = 75
english = 85
computer = 90
history = 70

# Calculate total and average
total = maths + science + english + computer + history
average = total / 5

# Display results
print("Total Marks:", total)
print("Average Marks:", average)
```
